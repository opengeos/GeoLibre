# Show or hide parts of the GeoLibre UI

`geolibre.Map` embeds the full GeoLibre app. Four options show or hide single pieces of its chrome:

| Argument | Hides or shows | Environment variable |
| --- | --- | --- |
| `toolbar` | Top toolbar and its menus | `GEOLIBRE_TOOLBAR` |
| `layer_panel` | Layers panel | `GEOLIBRE_LAYER_PANEL` |
| `style_panel` | Style (symbology) panel | `GEOLIBRE_STYLE_PANEL` |
| `status_bar` | Bottom status bar | `GEOLIBRE_STATUS_BAR` |

Each takes `True` (show), `False` (hide), or `None` (the default: keep the app's default). `app_layout`, `theme`, and the rendering engine (`renderer`) can also come from `GEOLIBRE_APP_LAYOUT`, `GEOLIBRE_THEME`, and `GEOLIBRE_RENDERER`.

```python
# %pip install -U geolibre
```

```python
from geolibre import Map

vector_url = "https://data.source.coop/giswqs/opengeos/world_cities.geojson"
raster_url = "https://data.source.coop/giswqs/opengeos/nlcd_2021_land_cover_30m.tif"
```

## Full UI with every part visible

The `full` layout shows the desktop chrome: toolbar, Layers and Style panels, and the status bar.

```python
m = Map(
    center=(-100, 40),
    zoom=3,
    app_layout="full",
    toolbar=True,
    layer_panel=True,
    style_panel=True,
    status_bar=True,
)
m.add_vector(vector_url, name="World Cities", circleRadius=3, fillColor="#ef4444")
m
```

## Menus only, no side panels

Keep the toolbar for adding data and running tools, but give the map the full width.

```python
m = Map(center=(-100, 40), zoom=3, app_layout="full", layer_panel=False, style_panel=False)
m.add_raster(raster_url, name="NLCD 2021")
m
```

## Layers panel only

Hide the toolbar, the Style panel, and the status bar, leaving a clean viewer with a layer list.

```python
m = Map(
    center=(-100, 40),
    zoom=3,
    app_layout="full",
    toolbar=False,
    layer_panel=True,
    style_panel=False,
    status_bar=False,
)
m.add_raster(raster_url, name="NLCD 2021")
m.add_vector(vector_url, name="World Cities", circleRadius=3, fillColor="#ef4444")
m
```

## Change the UI after the map is shown

The options are also widget traits. Setting one reloads the app with the new chrome and replays the project into it, so your layers stay. Run these one at a time and watch the map above.

```python
m.style_panel = True
```

```python
m.toolbar = True
m.status_bar = True
```

```python
print(m.toolbar, m.layer_panel, m.style_panel, m.status_bar)
```

## Set defaults with environment variables

Environment variables let a deployment (a JupyterHub image, a classroom server) pick the look of every map without changing notebook code. They are read when a `Map` is created, and an explicit argument always wins. Values such as `true`/`false`, `1`/`0`, `on`/`off`, `yes`/`no`, and `show`/`hide` are accepted.

You would usually set them before starting Jupyter (`export GEOLIBRE_LAYER_PANEL=hide`); here they are set from Python for the demo.

```python
import os

os.environ["GEOLIBRE_APP_LAYOUT"] = "full"
os.environ["GEOLIBRE_THEME"] = "dark"
os.environ["GEOLIBRE_STYLE_PANEL"] = "hide"
os.environ["GEOLIBRE_STATUS_BAR"] = "hide"
os.environ["GEOLIBRE_LAYER_PANEL"] = "hide"

m = Map(center=(-100, 40), zoom=3, layer_panel=True)  # beats GEOLIBRE_LAYER_PANEL
m.add_vector(vector_url, name="World Cities", circleRadius=3, fillColor="#facc15")
m
```

```python
print(m.app_layout, m.theme, m.toolbar, m.layer_panel, m.style_panel, m.status_bar)
```

Clear the variables so later maps in this session use the defaults again.

```python
for name in (
    "GEOLIBRE_APP_LAYOUT",
    "GEOLIBRE_THEME",
    "GEOLIBRE_STYLE_PANEL",
    "GEOLIBRE_STATUS_BAR",
    "GEOLIBRE_LAYER_PANEL",
):
    os.environ.pop(name, None)
```

## Switch rendering engines

The same project can be drawn by MapLibre (the default), Cesium (a 3D globe), Mapbox, or the ArcGIS Maps SDK. Pick one with `renderer=` when creating the map, or switch a displayed map with `set_renderer()`; layers carry over. Mapbox needs a Mapbox access token, set in the app's settings.

```python
m = Map(center=(-100, 40), zoom=3, app_layout="full", renderer="cesium", style_panel=False)
m.add_vector(vector_url, name="World Cities", circleRadius=3, fillColor="#ef4444")
m
```

```python
m.get_renderer()
```

```python
m.set_renderer("maplibre")  # back to the 2D map; try "arcgis" too
```

`GEOLIBRE_RENDERER` sets the default engine for every new map, like the UI variables above:

```python
os.environ["GEOLIBRE_RENDERER"] = "cesium"
print(Map().get_renderer())
os.environ.pop("GEOLIBRE_RENDERER")
```

## Export with the same chrome

`to_html` carries these settings, so the exported page looks like the map above. It loads the hosted GeoLibre app, which picks up the new options once that deployment includes this release.

```python
m.to_html("ui_chrome_map.html", title="GeoLibre UI options")
```
